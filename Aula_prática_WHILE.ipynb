49. Escreva um programa que imprima a sequência de números ímpares entre 1 e 101 (ambos incluídos), mas com os números divisíveis por 7 ou terminados em 7 substituídos por um asterisco (“*”). Veja a sequência abaixo. Cada número deve ser impresso em uma linha separada e nada mais deve ser impresso.

```python
numero = 1
while(numero <= 101):
    # primeira condicao = se o numero é divisil por 7
    # segunda condicao = se o numero termina com 7
    if((numero % 7 == 0) or (numero % 10 == 7)):
      print("*")
    else:
      print(numero)

    numero = numero + 2
```

50. Escreva um programa que imprima a sequência de números ímpares entre 1 e 101 (ambos incluídos), mas com os números divisíveis por 5 mas não divisíveis por 3 substituídos por um asterisco (“*”). Veja a sequência abaixo. Cada número deve ser impresso em uma linha separada e nada mais deve ser impresso.

```python
numero = 1
while(numero <= 101):
    # primeira condicao = se o numero é divisil por 5
    # segunda condicao = se o numero é divisil por 3
    if((numero % 5 == 0) and (numero % 3 == 0)):
      print(numero)
    else:
      if(numero % 5 == 0):
        print("*")
      else:
        print(numero)

    numero = numero + 2
```

Escreva um programa que enquanto o usuário não digitar o valor 0 (zero), continue somando todos os números que foram digitados. Caso o usuário digite 0 (zero), o programa deve imprimir a soma de todos os números e parar

```python
numero = 1
soma = 0
while(numero != 0):
  numero = int(input())

  if(numero != 0):
    soma = soma + numero


print(soma)
```
