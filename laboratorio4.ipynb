```python
#código que recebe 2 números e possiu condições para operações matemáticas
#uso de match case para cada opção matematica.
#
n1 = int(input("Digite nota 1.:"))
n2 = int(input("Digite nota 2.:"))
print("-" *17)
print("1- Soma das notas\n2- Média Simples\n3- Média Ponderada\n4- Maior nota\n5- Menor nota\n6- Diferença absoluta entre as notas\n7- Percentual de aproveitamento\n8- Situação do aluno (se aprovado)")
operacao = int(input(f"\nEscolha operação a ser realizada com os números {n1} e {n2}.:"))

match operacao:
  case 1:
    soma = n1 + n2
    print(f"\nO resultado da soma de {n1} e {n2} é {soma}")
  case 2:
    media = n1 + n2/2
    print(f"\nO resultado da média de {n1} e {n2} é {media}")
  case 3:
    peso1 = int(input("Digite o peso da nota 1.:"))
    peso2 = int(input("Digite o peso da nota 2.:"))
    ponderada = (n1*peso1) + (n2*peso2)/ (peso1 + peso2)
    print(f"\nO resultado da média ponderada de {n1} e {n2} é {ponderada}")
  case 4:
    if n1 > n2:
      print(f"\nO maior número é {n1}")
    else:
      print(f"\nO maior número é {n2}")
  case 5:
    if n1 > n2:
      print(f"\nO menor número é {n2}")
    else:
      print(f"\nO menor número é {n1}")
  case 6:
    if n1 > n2:
      diferenca = n1 - n2
    else:
      diferenca = n2 - n1
    print(f"\nA diferença absoluta é: {diferenca}")
  case 7:
    med = ((n1 + n2)/2)*1
    print(f"\nAproveitamento é de {med}%")
  case 8:
    if ((n1+n2)/2) > 70:
      print("\nAluno aprovado, atingiu média 70")
    else:
      print("\nAluno reprovado, não atingiu a média 70")
  case _:
    print("\nOpção inválida !!")

```

```python
#código que usa match case para ajustar cada entrada específica uma ação diferente.
#cenarios diferentes para forma de pagamento escolhida
valor = float(input("Digite o valor do curso.: "))
print("\nFormas de pagamento \n1-Cartão de Crédito \n2-Boleto \n3-Transferência \n")
pagamento = int(input("Digite a forma de pagamento escolhida.: "))

match pagamento:
  case 1:
    parcela = int(input("\nEscolha o número de parcelas, em até 3x sem juros.:"))
    if parcela >= 4:
      juros = (parcela - 3) * 0.02
      valor += valor * juros
      valor_parcela = valor / parcela
      print(f"Total do curso R$ {valor:.2f} conforme selecionado. Em {parcela} parcelas de R${valor_parcela:.2f}")
    else:
      valor_parcela = valor / parcela
      print(f"Total do curso R$ {valor:.2f} conforme selecionado. Em {parcela} parcelas de R${valor_parcela:.2f} ")
  case 2:
    desconto = valor * (10/100)
    desconto = valor - desconto
    print(f"Total do curso R$ {desconto:.2f} conforme selecionado")
  case 3:
    desconto = valor * (10/100)
    desconto = valor - desconto
    print(f"Total do curso R$ {desconto:.2f} conforme selecionado.")
  case _:
    print("Opção escolhida é inválida, necessário escolher entre [1-3]")
```
